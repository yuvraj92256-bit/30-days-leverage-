```python

import json
import requests
from google import genai
from google.colab import userdata



# 1. Initialize client
client = genai.Client(api_key=userdata.get("GEMINI_API_KEY"))

# 2. Ingest raw API data
url = "https://jsonplaceholder.typicode.com/posts"
raw_posts = requests.get(url).json()[:4]

processed_records = []

# 3. Pipeline execution
for post in raw_posts:
    prompt = f"""
    Analyze this post:
    Title: {post['title']}
    Body: {post['body']}

    Return ONLY a valid JSON object with these exact keys:
    - category (tech, business, or lifestyle)
    - urgency (low, medium, or high)
    - summary (one concise sentence)
    """

    response = client.models.generate_content(
        model="gemini-2.5-flash",
        contents=prompt,
        config={"response_mime_type": "application/json"}
    )

    parsed_data = json.loads(response.text)

    # Merge original metadata with AI analysis
    record = {
        "post_id": post["id"],
        "title": post["title"],
        **parsed_data
    }
    processed_records.append(record)

# 4. Display output
print(json.dumps(processed_records, indent=2))
```

```python
# 1. Define the blueprint (notice the 4 spaces inside)
def calculate_conversion(calls_made, meetings_booked):
    conversion_rate = (meetings_booked / calls_made) * 100

    # 'return' is how the tool hands the final math back to the main timeline
    return conversion_rate

# 2. The main timeline (0 spaces). The tool is built. Now we use it.
monday_rate = calculate_conversion(50, 2)
tuesday_rate = calculate_conversion(80, 5)
wednesday_rate = calculate_conversion(100,6)

print(f"Monday Conversion: {monday_rate}%")
print(f"Tuesday Conversion: {tuesday_rate}%")
print(f"Wednesday Conversion: {wednesday_rate}%")
```

```python
import requests
import time
from datetime import datetime

print("---SOME AUTOMATION SHIT---")

while True:
    url = "https://api.frankfurter.app/latest?from=USD&to=INR"
    response = requests.get(url)
    live_data = response.json()
    exchange_rate = live_data["rates"]["INR"]

    now = datetime.now().strftime("%Y-%m-%d %H:%M:%S")

    log_entry = f"{now} USD TO INR: {exchange_rate}\n"

    with open("currency.log.txt" , "a") as file:
        file.write(log_entry)
    print(f"Successfully Logged: {log_entry.strip()}")
    time.sleep(10)
```
