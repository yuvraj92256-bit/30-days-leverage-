```python
incoming_leads = [
    {"sender": "Dr. Sharma", "text": "Our clinic website crashed and we are losing 20 patients a day. Need an urgent automated fix today, budget isn't an issue."},
    {"sender": "Rahul V.", "text": "Hey bro can you make me a free clone of Instagram by tonight? Will give you shoutout."},
    {"sender": "Apex Diagnostics", "text": "We are looking to upgrade our patient booking pipeline next month. Send over your portfolio and standard pricing."}
]

prompt_template = """
You are an automated backend classifier. Analyze this inbound message:
"{text}"

Classify it into strictly ONE of these categories:
- HOT_LEAD (has urgent need, ready to pay)
- FOLLOW_UP (potential project, not urgent)
- TRASH (unrealistic, free work, low intent)

Respond with ONLY the category name and a 5-word reason. Format: [CATEGORY] - [Reason]
"""

for lead in incoming_leads:
    response = client.models.generate_content(
        model="models/gemini-3.6-flash",
        contents=prompt_template.format(text=lead["text"])
    )
    print(f"Lead: {lead['sender']} -> {response.text.strip()}")
```

```python
import requests
url = "https://jsonplaceholder.typicode.com/posts](https://jsonplaceholder.typicode.com/posts"
response = requests.get(url)
posts = response.json()
for item in posts[:3]:
    prompt = f"Analyze this post.\nTitle: {item['title']}\nBody: {item['body']}\nOutput strictly: [URGENCY: HIGH/LOW] - [5-word summary]"
    ai_response = client.models.generate_content(
        model="models/gemini-3.6-flash",
        contents=prompt)
print(f"Post #{item['id']} -> {ai_response.text.strip()}")
```
