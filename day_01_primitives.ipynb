```python
raw_scores = [45, 82, 91, 33, 76]
# Filter out low-value items
qualified = [score for score in raw_scores if score >= 75]
print("Qualified leads count:", len(qualified))
```

```python
import requests

# 1. Reach out to a live web server
response = requests.get("https://httpbin.org/json")

# 2. Convert the server's response into a Python dictionary
data = response.json()

# 3. Extract and print specific pieces of that data
print("Status Code:", response.status_code)
print("Title:", data["slideshow"]["title"])
print("Author:", data["slideshow"]["author"])
```
